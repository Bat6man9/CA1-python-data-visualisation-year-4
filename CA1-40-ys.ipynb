```python
# Define a function to clean whole-number employment and annual wage estimates
def make_int(i):

    try:
        # Convert a valid estimate into a whole number
        return int(i)

    except:
        # Record non-numerical entries, such as "*", as missing values
        return np.nan
```

```python
# Apply make_int to each employment estimate in TOT_EMP
# Valid whole-number estimates become numerical values
# Entries that cannot be converted are recorded as NaN
df["TOT_EMP"] = df["TOT_EMP"].map(make_int)

# Display the first five employment estimates after conversion
df["TOT_EMP"].head()
```

```python
# Count the TOT_EMP entries recorded as NaN after numerical conversion
# This includes non-numerical entries that make_int could not convert
df["TOT_EMP"].isnull().sum()
```

```python
# Inspect the first ten annual mean wage estimates in A_MEAN
# Check their values and current data type before numerical conversion
df["A_MEAN"].head(10)
```

```python
# Apply make_int to each annual mean wage estimate in A_MEAN
# Convert valid whole-dollar estimates into numerical values
# Record entries that cannot be converted as NaN
df["A_MEAN"] = df["A_MEAN"].map(make_int)

# Display the first five annual mean wages after conversion
df["A_MEAN"].head()
```

```python
# Count missing annual mean wage estimates in A_MEAN after conversion
# This includes entries that make_int could not convert into numbers
df["A_MEAN"].isnull().sum()
```

```python
# Count rows with identical values across every column
# Repeated occupation names alone do not make rows duplicates
df.duplicated().sum()
```

Data cleaning checks

TOT_EMP contains employment estimates, while A_MEAN contains mean annual wage estimates. I converted these columns into numbers so they can be used for calculations and charts.

Entries that could not be converted were recorded as NaN. After conversion, TOT_EMP had 11,598 missing values and A_MEAN had 5,352. These do not mean zero employment or zero wages. Some wage symbols represent a published limit rather than an exact amount.

I also checked for rows with identical values across every column. The result was zero, so no duplicate rows needed to be removed.

These checks cover the two columns converted so far. Other columns will be checked before they are used. For each chart, I will leave out only the records missing the values that chart needs.

```python
# Check the data types of the employment, wage and concentration measures
# Confirm the converted columns are numerical and identify any remaining object columns
df[["TOT_EMP", "A_MEAN", "JOBS_1000", "LOC_QUOTIENT"]].dtypes
```

```python
# Define a function to convert employment rates and location quotients into decimal numbers
def make_float(i):
    try:
        # Convert a numerical estimate while preserving its decimal places
        return float(i)
    except:
        # Record entries that cannot be converted, such as "**", as missing values
        return np.nan
```

```python
# Convert employment per 1,000 jobs into numbers while keeping decimal places
df["JOBS_1000"] = df["JOBS_1000"].map(make_float)

# Convert location quotients into numbers while keeping decimal places
df["LOC_QUOTIENT"] = df["LOC_QUOTIENT"].map(make_float)

# Check that both columns now have numerical data types
df[["JOBS_1000", "LOC_QUOTIENT"]].dtypes
```

```python
# Count missing values in the two measures after decimal conversion
# Include both existing blanks and entries that could not be converted
df[["JOBS_1000", "LOC_QUOTIENT"]].isnull().sum()
```

```python
# Store the occupation codes selected for this analysis
occupation_codes = [
    "15-1221",  # Computer and Information Research Scientists
    "15-1252",  # Software Developers
    "15-2031",  # Operations Research Analysts
    "15-2041",  # Statisticians
    "15-2051"   # Data Scientists
]

# Display the selected occupation codes
occupation_codes
```

Occupations selected for the analysis

I selected Computer and Information Research Scientists, Software Developers, Operations Research Analysts, Statisticians and Data Scientists.

These occupations allow me to compare employment and wages across computing research, software development, data science and related analytical work.

This is a selected group, rather than a complete list of every occupation connected to AI. The dataset does not tell me which workers within each occupation actually work with AI. For example, a software developer may work on software that does not involve AI.

I will therefore describe the results as employment in these selected occupations, rather than claiming they show the exact number of AI workers.

```python
# Copy the dataframe so the following filters do not change df
# Store the records being prepared for the charts in filtered_df
filtered_df = df.copy()

# Keep employment and wage estimates calculated across all industries
# Exclude estimates for individual industries to avoid mixing different coverage
filtered_df = filtered_df[filtered_df["I_GROUP"] == "cross-industry"]

# Keep ownership code 1235, which combines private and government ownership
# Exclude separate ownership categories so they are not mixed with the combined estimates
filtered_df = filtered_df[filtered_df["OWN_CODE"] == 1235]

# Keep individual occupations identified by the detailed occupation category
# Exclude broader groups and totals that could overlap with those occupations
filtered_df = filtered_df[filtered_df["O_GROUP"] == "detailed"]

# Display the first five remaining occupation names and their category values
# Check that the displayed records match the three filters applied above
filtered_df[["OCC_TITLE", "I_GROUP", "OWN_CODE", "O_GROUP"]].head()
```

```python
# Select the records matching each code in occupation_codes
# Combine those records into one dataframe containing the five selected occupations
occupation_df = pd.concat([
    filtered_df[filtered_df["OCC_CODE"] == code]
    for code in occupation_codes
])

# Preview the first five selected records, including their areas and numerical estimates
# National and local records are still included and will be separated next
occupation_df[["AREA_TITLE", "OCC_CODE", "OCC_TITLE", "TOT_EMP", "A_MEAN"]].head()
```

```python
# Keep national estimates, identified by AREA_TYPE 1
# Store them separately for comparisons between occupations across the USA
national_df = occupation_df[occupation_df["AREA_TYPE"] == 1].copy()

# Keep state-level estimates, identified by AREA_TYPE 2
# Include states and Washington, DC, while excluding territories and smaller local areas
state_df = occupation_df[occupation_df["AREA_TYPE"] == 2].copy()

# Display the national employment and annual mean wage estimates
# Check that the output contains one national record for each selected occupation
national_df[["AREA_TITLE", "OCC_CODE", "OCC_TITLE", "TOT_EMP", "A_MEAN"]]
```

```python
# Count the records retained at each geographical level
# National data should contain one record for each of the five occupations
print("National records:", national_df.shape[0])
print("State-level records:", state_df.shape[0])

# Check whether a selected occupation appears more than once nationally
print("Repeated national occupations:",
      national_df[["OCC_CODE"]].duplicated().sum())

# Check whether the same occupation appears more than once within the same state
# Different occupations within one state are valid separate records
print("Repeated state-occupation records:",
      state_df[["AREA", "OCC_CODE"]].duplicated().sum())

# Count missing employment and wage estimates in the national selection
print("Missing national estimates:")
print(national_df[["TOT_EMP", "A_MEAN"]].isnull().sum())

# Count missing values in the four measures used for state comparisons
# These counts determine which records need excluding from individual charts
print("Missing state-level estimates:")
print(state_df[["TOT_EMP", "A_MEAN", "JOBS_1000", "LOC_QUOTIENT"]].isnull().sum())
```
